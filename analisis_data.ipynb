```python
import pandas as pd
import numpy as np

np.random.seed(42)

kota_list = ['Jakarta', 'Bandung', 'Surabaya', 'Yogyakarta', 'Medan']
kondisi_list = ['Cerah', 'Berawan', 'Hujan', 'Hujan Lebat']

data = []
dates = pd.date_range(start='2024-01-01', periods=200, freq='D')

for tanggal in dates:
    for kota in kota_list:
        # Karakteristik tiap kota
        if kota == 'Bandung':
            suhu = round(np.random.uniform(18.0, 26.0), 1)
            kelembaban = np.random.randint(70, 98)
        elif kota == 'Jakarta':
            suhu = round(np.random.uniform(28.0, 35.5), 1)
            kelembaban = np.random.randint(60, 88)
        else:
            suhu = round(np.random.uniform(23.0, 33.0), 1)
            kelembaban = np.random.randint(65, 92)
            
        kondisi = np.random.choice(kondisi_list, p=[0.4, 0.3, 0.2, 0.1])
        
        data.append({
            'Tanggal': tanggal,
            'Kota': kota,
            'Suhu (°C)': suhu,
            'Kelembaban (%)': kelembaban,
            'Kondisi': kondisi
        })

df = pd.DataFrame(data)
df.to_csv('dataset_cuaca.csv', index=False)
print("✅ SUCCESS: 1.000 Data Sintetis Cuaca Berhasil Dibuat!")
```
